```python
# Import pandas - 'pd' is the conventional alias
import pandas as pd
```

```python
# Build a Series from a list - pandas supplies the default 0,1,2 index
data = [100,102,104]
series = pd.Series(data)
print(series)
```

```python
# Same values, but with our own labels as the index
series = pd.Series(data,index=["a","b","c"])
print(series)
```

```python
print(series.loc["a"])
```

```python
# Assign by label with .loc
series.loc["c"]=500
print(series)
```

```python
print(series.iloc[2])
```

```python
print(series[series < 200])
```

```python
# Assigning to a label that doesn't exist yet appends a new entry
series.loc["d"]= 509
print(series)
```

```python
# A plain dict: day -> calories burnt
calories = {
    "Day 1": 1750,
    "Day 2": 2500,
    "Day 3": 3000,
    "Day 4": 1700,
    "Day 5": 1650
}
```

```python
# Series from a dict - the keys become the index
series = pd.Series(calories)
print(series)
```

```python
# Careful: an index that doesn't match the dict keys gives all NaN
series = pd.Series(calories,index=["a","b","c","d","e"])
print(series)
```

```python
# DataFrame input is a dict of column name -> list of values
data = {
    "name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age": [25, 30, 35, 40, 45]
}
```

```python
# Build the DataFrame, naming each row instead of using 0,1,2
df = pd.DataFrame(data,index=["Employee 1", "Employee 2", "Employee 3", "Employee 4", "Employee 5"])
print(df)
```

```python
print(df.loc["Employee 3"])
```

```python
print(df.iloc[0])
```

```python
print(df["name"])
```

```python
# Add a whole new column by assigning a list
df["Job"]=["Engineer", "N/A", "Analyst", "Clerk", "HR"]
print(df)
```

```python
# Append more rows by concatenating a second DataFrame
new_rows = pd.DataFrame([
    {"name": "Frank", "Age": 50, "Job": "Consultant"},
    {"name": "Grace", "Age": 28, "Job": "N/A"}
], index=["Employee 6", "Employee 7"])
df = pd.concat([df, new_rows])
print(df)
```
