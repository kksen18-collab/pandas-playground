```python
# Re-read the Pokemon data, using the Name column as the index
df = pd.read_csv("data/pokemon.csv",index_col="Name")
```

```python
print(df)
```

```python
print(df.loc["Pikachu",["Attack","Defense"]])
```

```python
print(df[df["Speed"]>60])   
```

```python
print(df.loc[df["Legendary"]==True].index)
```

```python
# Group the rows by primary type - this object is lazy until we aggregate
group = df.groupby("Type 1")
print(group)
```

```python
# Lowest Speed within each type group
group["Speed"].min()
```

```python
# How many Pokemon fall into each type group
group["Speed"].count()
```
