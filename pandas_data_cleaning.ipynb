```python
# Switch to the second dataset - exercise logs
df_data = pd.read_csv("data/pandas_practice_dataset.csv")
print(df_data)
```

```python
# What is the mean of calories burnt based on duration
df_data.groupby("Duration")["Calories"].mean()
```

```python
# Correct Incorrect data format and set appropriate data types - handle NaN values too

print(df_data["Date"])
```

```python
# Strip the stray apostrophes out of the date strings
df_data["Date"]=df_data["Date"].str.replace("'","")
print(df_data["Date"])
```

```python
# Parse the YYYY/MM/DD dates; anything that doesn't match becomes NaT
format_slashes = pd.to_datetime(df_data["Date"], format="%Y/%m/%d", errors="coerce")
print(format_slashes)
```

```python
# Parse the YYYYMMDD dates; anything that doesn't match becomes NaT
format_numeric = pd.to_datetime(df_data["Date"], format="%Y%m%d", errors="coerce")
print(format_numeric)
```

```python
# Combine both attempts so each row keeps whichever format parsed
df_data["Date"]=format_slashes.fillna(format_numeric)
print(df_data["Date"])
```

```python
# Drop rows whose date matched neither format
df_data=df_data.dropna(subset=["Date"])
```

```python
print(df_data)
```

```python
# See which other columns still have missing values
df_data.isnull().sum()
```

```python
# Drop the remaining rows with NaNs
df_data = df_data.dropna()
```

```python
print(df_data)
```

```python
# Per duration: how many days were logged and the lowest calorie burn
df_data.groupby("Duration").agg(number_of_days=("Date", "count"),
                                min_calories=("Calories", "min"))
```
